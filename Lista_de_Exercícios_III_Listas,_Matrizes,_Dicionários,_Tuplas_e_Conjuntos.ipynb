```python
def primo(n):
    if n < 2: return False
    for i in range(2, n):
        if n % i == 0: return False
    return True

nums = [int(input("Número: ")) for _ in range(10)]
primos = [n for n in nums if primo(n)]
print(primos)
```

```python
mat = [[int(input()) for _ in range(3)] for _ in range(3)]
identidade = True
for i in range(3):
    for j in range(3):
        if (i == j and mat[i][j] != 1) or (i != j and mat[i][j] != 0):
            identidade = False
print("Identidade" if identidade else "Não é identidade")
```

```python
produtos = {
    input("Produto: "): [float(input("Preço: ")) for _ in range(3)]
    for _ in range(3)
}

maior = None
nome_maior = ""
for nome, precos in produtos.items():
    dif = max(precos) - min(precos)
    if maior is None or dif > maior:
        maior = dif
        nome_maior = nome

print("Maior variação:", nome_maior)
```
